## Concept

A DAG expresses ordering constraints. Independent tasks in the same level are parallel-eligible, not automatically parallel.

```python
from pyworkflowkit._compat.v1_root import WorkflowRuntime
from pyworkflowkit.control_plane import WorkflowRuntimeProvider

workflow_definition: dict[str, object] = {
    "workflow_id": "notebooks.dag",
    "version": "1",
    "tasks": [
        {"task_id": "prepare", "handler_ref": "handlers:prepare"},
        {
            "task_id": "build_api",
            "handler_ref": "handlers:build_api",
            "depends_on": ["prepare"],
        },
        {
            "task_id": "build_ui",
            "handler_ref": "handlers:build_ui",
            "depends_on": ["prepare"],
        },
        {
            "task_id": "publish",
            "handler_ref": "handlers:publish",
            "depends_on": ["build_api", "build_ui"],
        },
    ],
}
```

```python
inspection = WorkflowRuntimeProvider(WorkflowRuntime()).inspect_workflow(workflow_definition)
for group in inspection.groups:
    print({"level": group.index, "tasks": list(group.task_ids)})
```

Add a dependency from `build_ui` to `build_api` and observe how the execution groups change.
