## Construction

Register handlers explicitly, execute a definition, then inspect runtime evidence.

```python
from pyworkflowkit._compat.v1_root import TaskHandle, WorkflowRuntime, task, workflow


@task
def hello() -> str:
    return "hello"


@workflow(id="notebooks.runtime", version="1")
def demo() -> tuple[TaskHandle, ...]:
    return (hello,)


runtime = WorkflowRuntime()
runtime.register(hello.handler_ref, hello.handler)
definition = demo.build()
```

```python
run = runtime.run(definition)
print({"status": run.status.value, "event_count": len(runtime.events(run.run_id))})
```

## Observation

The runtime owns execution state and evidence; the task handler stays focused on workload logic.
