## Construction

Run two dependent tasks and inspect the ordered runtime facts they emit.

```python
from pyworkflowkit import TaskHandle, WorkflowRuntime, task, workflow


@task
def prepare() -> str:
    return "ready"


@task(depends_on=(prepare,))
def publish() -> str:
    return "published"


@workflow(id="notebooks.events", version="1")
def demo() -> tuple[TaskHandle, ...]:
    return (prepare, publish)
```

```python
runtime = WorkflowRuntime()
for handle in demo.task_handles():
    runtime.register(handle.handler_ref, handle.handler)

run = runtime.run(demo.build())
events = runtime.events(run.run_id)
for event in events:
    print(event.event_sequence, event.event_type.value)
```

## Observation

Events are immutable runtime facts. They are not a substitute for mutable current state.
