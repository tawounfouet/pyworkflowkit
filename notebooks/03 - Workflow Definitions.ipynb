## Construction

Compose task handles into an immutable `WorkflowDefinition`.

```python
from pyworkflowkit._compat.v1_root import TaskHandle, task, workflow


@task
def prepare() -> str:
    return "ready"


@task(depends_on=(prepare,))
def publish() -> str:
    return "published"


@workflow(id="notebooks.definition", version="1", description="Definition exploration")
def demo() -> tuple[TaskHandle, ...]:
    return (prepare, publish)
```

```python
definition = demo.build()
print(
    {
        "workflow_id": str(definition.workflow_id),
        "version": definition.version,
        "task_count": len(definition.tasks),
    }
)
```

```python
for task_definition in definition.tasks:
    print(str(task_definition.task_id), [str(dep) for dep in task_definition.depends_on])
```
