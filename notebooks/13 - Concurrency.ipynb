## Concept

Planning reveals concurrency opportunities. Execution strategy remains a separate runtime concern.

```python
from pyworkflowkit._compat.v1_root import WorkflowRuntime
from pyworkflowkit.control_plane import WorkflowRuntimeProvider

workflow_definition: dict[str, object] = {
    "workflow_id": "notebooks.concurrency",
    "version": "1",
    "tasks": [
        {"task_id": "prepare", "handler_ref": "handlers:prepare"},
        {
            "task_id": "build_api",
            "handler_ref": "handlers:build_api",
            "depends_on": ["prepare"],
        },
        {
            "task_id": "build_ui",
            "handler_ref": "handlers:build_ui",
            "depends_on": ["prepare"],
        },
        {
            "task_id": "publish",
            "handler_ref": "handlers:publish",
            "depends_on": ["build_api", "build_ui"],
        },
    ],
}
```

```python
inspection = WorkflowRuntimeProvider(WorkflowRuntime()).inspect_workflow(workflow_definition)
parallel_groups = [list(group.task_ids) for group in inspection.groups if len(group.task_ids) > 1]
print({"parallel_eligible_groups": parallel_groups})
```

## Experiment

Make one branch depend on the other and observe how the parallel-eligible group disappears.
