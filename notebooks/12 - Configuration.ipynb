## Construction

Build explicit runtime settings and inspect the normalized configuration.

```python
from pathlib import Path

from pyworkflowkit._compat.v1_root import RuntimeSettings

settings = RuntimeSettings.load(
    overrides={
        "runtime": {"workspace": Path(".pyworkflow-notebook")},
        "metadata": {
            "backend": "sqlite",
            "sqlite_path": "notebook.sqlite3",
            "sqlite_wal": False,
        },
    }
)
```

```python
print(
    {
        "workspace": str(settings.runtime.workspace),
        "backend": settings.metadata.backend,
        "sqlite_path": str(settings.metadata.sqlite_path),
    }
)
```
