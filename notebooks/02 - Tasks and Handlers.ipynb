## Construction

A `@task` produces a reusable task handle that carries identity, handler reference, and execution metadata.

```python
from pyworkflowkit import task


@task(description="Return a deterministic row count.")
def fetch() -> dict[str, int]:
    return {"rows": 3}
```

```python
definition = fetch.to_definition()
print(
    {
        "task_id": str(fetch.task_id),
        "handler_ref": fetch.handler_ref,
        "description": definition.description,
    }
)
```
