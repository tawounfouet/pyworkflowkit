This notebook follows the DX05 loop: **Concept → Construction → Inspection → Modification → Execution → Observation → Experiment**.

## Construction

Build the smallest public-API workflow, then inspect the definition before execution.

```python
from pyworkflowkit import TaskHandle, WorkflowRuntime, task, workflow


@task
def hello() -> str:
    return "hello"


@workflow(id="notebooks.hello", version="1")
def demo() -> tuple[TaskHandle, ...]:
    return (hello,)
```

```python
definition = demo.build()
print({"workflow_id": str(definition.workflow_id), "tasks": [str(task.task_id) for task in definition.tasks]})
```

## Execution and observation

```python
runtime = WorkflowRuntime()
runtime.register(hello.handler_ref, hello.handler)
run = runtime.run(definition)
print({"status": run.status.value, "events": len(runtime.events(run.run_id))})
```
