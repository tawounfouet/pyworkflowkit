## Construction

Author against the public ecosystem SDK and inspect the resulting plugin descriptor.

```python
from pyworkflowkit.ecosystem import PluginType, RegisteredPlugin, plugin_registration


class ExamplePlugin:
    name = "notebook-example"


def plugin() -> RegisteredPlugin[ExamplePlugin]:
    return plugin_registration(
        name="notebook-example",
        plugin_type=PluginType.WORKLOAD,
        factory=ExamplePlugin,
        plugin_version="0.1.0",
        description="Interactive DX05 plugin example.",
    )
```

```python
registration = plugin()
instance = registration.create()
print(
    {
        "name": registration.descriptor.name,
        "type": registration.descriptor.plugin_type.value,
        "instance": type(instance).__name__,
    }
)
```
