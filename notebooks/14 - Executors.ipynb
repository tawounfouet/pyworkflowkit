## Concept

Executors declare capabilities explicitly. The coordinator uses those contracts instead of assuming every executor behaves the same.

```python
from pyworkflowkit.ecosystem import (
    CancellationCapability,
    ExecutorCapabilities,
    TimeoutCapability,
)

sequential = ExecutorCapabilities()
cooperative = ExecutorCapabilities(
    supports_parallelism=True,
    timeout=TimeoutCapability.SOFT,
    cancellation=CancellationCapability.COOPERATIVE,
    max_concurrency=4,
)
```

```python
for name, capabilities in {
    "sequential": sequential,
    "cooperative": cooperative,
}.items():
    print(
        name,
        {
            "parallel": capabilities.supports_parallelism,
            "timeout": capabilities.timeout.value,
            "cancellation": capabilities.cancellation.value,
            "max_concurrency": capabilities.max_concurrency,
        },
    )
```
